## Background: a dict keyed by pairs

A dict key can be a tuple: `{("Ash", "Birch"): 4}` means "the trip between Ash and Birch takes 4". `for (a, b), mins in trips.items():` unpacks the pair and the value in one go. If trips go both ways, each pair is a neighbor of the other.

```python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
trips = {("Ash", "Birch"): 4, ("Birch", "Cedar"): 3}
for (a, b), mins in trips.items():
    print(a, "<->", b, mins, "minutes")
```
