# Toolbox · "What's that thing called?"

A lookup table from **what you want to do** to **what Python calls it**, with the Linux command you
already know where there is one. Search this page with **Cmd+F** for a word like "split", "sort"
or "count". Every **Demo** cell runs as is; change it and re-run it to experiment.

Two kinds of tool appear below:
- **functions** you call with the value inside: `len(items)`, `int("3")`, `sorted(items)`;
- **methods** that belong to a value, written after a dot: `"a:b".split(":")`, `items.append(x)`.

## Text (`str`)

| I want to... | Python | Example → result | Like in Linux |
|---|---|---|---|
| cut text at a character | `s.split(":")` | `"a:b".split(":")` → `["a", "b"]` | `cut -d:` |
| cut at any run of spaces | `s.split()` | `"a   b".split()` → `["a", "b"]` | `awk` fields |
| cut once, from the right | `s.rsplit(":", 1)` | `"a:b:c".rsplit(":", 1)` → `["a:b", "c"]` | |
| glue pieces with a separator | `"-".join(pieces)` | `"-".join(["a", "b"])` → `"a-b"` | `paste -sd-` |
| trim spaces at both ends | `s.strip()` | `"  a ".strip()` → `"a"` | |
| test the start or the end | `s.startswith("#")`, `s.endswith(".log")` | `"#x".startswith("#")` → `True` | `grep '^#'` |
| test whether it contains | `"x" in s` | `":" in "a:b"` → `True` | `grep -q` |
| replace | `s.replace("a", "b")` | `"a-a".replace("-", ":")` → `"a:a"` | `sed 's/a/b/g'` |
| change case | `s.upper()`, `s.lower()` | `"milk".upper()` → `"MILK"` | `tr a-z A-Z` |
| count characters | `len(s)` | `len("apple")` → `5` | `wc -c` |
| one character, or a range | `s[0]`, `s[-1]`, `s[1:3]` | `"apple"[0:2]` → `"ap"` | `cut -c1-2` |
| put values into text | `f"{n} apples"` | with `n = 3` → `"3 apples"` | `printf` |
| pad a number with zeros | `f"{n:04d}"` | `f"{7:04d}"` → `"0007"` | `printf %04d` |

```python
# Demo: run this cell as it is (Shift+Enter). Change it and re-run to experiment.
line = "  2026-03-01 09:02 alice apple 3 0.50  "
print(line.split())
print(line.strip().startswith("2026"))
print("notes.final.txt".rsplit(".", 1))
print("-".join(["red", "green", "blue"]), f"{7:04d}", f"{3} apples")
```

## Numbers (`int`, `float`)

| I want to... | Python | Example → result |
|---|---|---|
| text to a whole number | `int(text)` | `int("3")` → `3` (raises `ValueError` if it isn't one) |
| text to a decimal number | `float(text)` | `float("1.99")` → `1.99` |
| a number to text | `str(n)` | `str(3)` → `"3"` |
| divide, whole part only | `a // b` | `1250 // 100` → `12` |
| the remainder | `a % b` | `1250 % 100` → `50` |
| power | `a ** b` | `2 ** 3` → `8` |
| the smaller / larger | `min(a, b)`, `max(a, b)` | `min(3, 10)` → `3` |
| round, or drop the sign | `round(x, 2)`, `abs(x)` | `round(1.2345, 2)` → `1.23` |

```python
# Demo: run this cell as it is (Shift+Enter). Change it and re-run to experiment.
print(int("3"), float("1.99"), str(3))
print(1250 // 100, 1250 % 100, 2 ** 3, min(3, 10), round(1.2345, 2))
```

## Lists (`list`): many things in order

| I want to... | Python | Example → result | Like in Linux |
|---|---|---|---|
| make one | `[]`, `[1, 2, 3]` | | |
| add to the end | `items.append(x)` | `[1].append(2)` makes it `[1, 2]` | `>>` |
| add several to the end | `items.extend(more)` | | |
| remove from the end | `items.pop()` | returns the removed item | |
| remove from the front | `items.pop(0)` | returns it (slow on huge lists; use a `deque`) | |
| one item by position | `items[0]`, `items[-1]` | first, last | `head -1`, `tail -1` |
| a range of items | `items[1:3]`, `items[:n]` | `[a, b, c, d][:2]` → `[a, b]` | `head -n` |
| reverse | `items[::-1]` | `[1, 2, 3][::-1]` → `[3, 2, 1]` | `tac` |
| how many | `len(items)` | | `wc -l` |
| is it in there | `x in items` | | `grep -q` |
| sorted copy | `sorted(items)` | | `sort` |
| sort by part of each item | `sorted(items, key=lambda x: x[1])` | sort pairs by their second part | `sort -k2` |
| largest first | `sorted(items, reverse=True)` | | `sort -r` |
| add them up | `sum(items)` | | `awk '{s+=$1} END {print s}'` |
| smallest / largest | `min(items)`, `max(items)` | | |
| a real copy | `list(items)` | `b = a` is NOT a copy | `cp` |
| number each item | `for i, x in enumerate(items):` | `i` is 0, 1, 2... | `nl` |
| walk two lists side by side | `for a, b in zip(xs, ys):` | | `paste` |

```python
# Demo: run this cell as it is (Shift+Enter). Change it and re-run to experiment.
basket = ["apple", "bread", "milk"]
basket.append("cheese")
print(basket[0], basket[-1], basket[:2], len(basket), "milk" in basket)
print(basket[::-1])
print(sorted([("bread", 325), ("apple", 50)], key=lambda pair: pair[1]))
for i, item in enumerate(basket):
    print(i, item)
```

## Dicts (`dict`): find by name

A dict is a lookup table: give it a **key**, get back its **value**. Like `declare -A` in bash.

| I want to... | Python | Note |
|---|---|---|
| make one | `{}`, `{"apple": 50}` | `{}` is an empty dict, not a set |
| read a value | `d["apple"]` | raises `KeyError` if the key is missing |
| read, with a default if missing | `d.get("apple", 0)` | never raises |
| add or overwrite | `d["apple"] = 55` | |
| is this key in it | `"apple" in d` | checks keys, not values |
| walk keys and values | `for key, value in d.items():` | |
| just the keys / values | `d.keys()`, `d.values()` | |
| remove a key | `del d["apple"]` | |
| count per key | `counts[k] = counts.get(k, 0) + 1` | `uniq -c`, by hand |
| a list per key | `if k not in groups: groups[k] = []` then `groups[k].append(x)` | or `groups.setdefault(k, []).append(x)` |

```python
# Demo: run this cell as it is (Shift+Enter). Change it and re-run to experiment.
prices = {"apple": 50}
prices["bread"] = 325
print(prices["bread"], prices.get("tea", 0), "apple" in prices)
counts = {}
for item in ["apple", "milk", "apple"]:
    counts[item] = counts.get(item, 0) + 1
print(counts)
for name, cents in prices.items():
    print(name, "->", cents)
```

## Sets (`set`) and tuples (`tuple`)

| I want to... | Python | Note |
|---|---|---|
| a collection with no repeats | `set()`, `{"a", "b"}` | `set()` for an empty one |
| add / remove one | `s.add(x)`, `s.discard(x)` | `discard` doesn't complain if it's missing |
| in both / in either / in one but not the other | `a & b`, `a \| b`, `a - b` | like `comm` |
| a fixed record that can't change | `pair = ("apple", 50)` | can be a dict key |
| split a record into names | `name, price = pair` | the number of names must match |
| return two values | `return a, b` | the caller gets a tuple |

## Building a list or dict in one line (comprehensions), and tiny functions

| I want to... | Python | Example → result |
|---|---|---|
| a new list from each item | `[x * 2 for x in items]` | `[1, 2]` → `[2, 4]` |
| keep only some items | `[x for x in items if x > 1]` | `[1, 2, 3]` → `[2, 3]` |
| a new dict | `{k: v * 2 for k, v in d.items()}` | `{"a": 1}` → `{"a": 2}` |
| a dict without one key | `{k: v for k, v in d.items() if k != "a"}` | |
| any one key of a dict | `next(iter(d))` | the first key |
| a tiny unnamed function | `lambda x: x[1]` | same as `def f(x): return x[1]` |
| flip a list in place | `items.reverse()` | changes `items` itself |

```python
# Demo: run this cell as it is (Shift+Enter). Change it and re-run to experiment.
prices = {"apple": 50, "bread": 325, "milk": 199}
print([p * 2 for p in prices.values()])
print([name for name, p in prices.items() if p > 100])
print({name: p for name, p in prices.items() if name != "bread"})
print(next(iter(prices)))
```

## Control flow: deciding and repeating

| I want to... | Python | Like in bash |
|---|---|---|
| decide | `if x > 0:` / `elif x == 0:` / `else:` | `if ...; then ... elif ... else ... fi` |
| combine conditions | `a and b`, `a or b`, `not a` | `&&`, `\|\|`, `!` |
| once per item | `for item in items:` | `for item in ...; do ... done` |
| n times | `for _ in range(n):` | `for i in $(seq n)` |
| as long as something is true | `while condition:` | `while ...; do ... done` |
| leave a loop early / skip to the next round | `break`, `continue` | `break`, `continue` |
| hand back an answer and stop | `return value` | |
| "no value" | `None`; test it with `x is None` | an empty variable |
| survive an error | `try:` ... `except ValueError:` ... | `cmd \|\| fallback` |

Indentation (4 spaces) is how Python knows what's inside an `if`, a loop, or a function, where
bash uses `then`/`fi` and `do`/`done`.

```python
# Demo: run this cell as it is (Shift+Enter). Change it and re-run to experiment.
for qty in [1, 5, 12]:
    if qty >= 10:
        print(qty, "bulk order")
    else:
        print(qty, "normal order")

for text in ["3", "abc"]:
    try:
        print(int(text))
    except ValueError:
        print(text, "is not a number")
```
