## Background: counting with a dict

`counts.get(word, 0)` reads the current count, or 0 if the word hasn't been seen yet. So `counts[word] = counts.get(word, 0) + 1` adds one, and it works the first time too.

```python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
counts = {}
for word in ["tea", "milk", "tea"]:
    counts[word] = counts.get(word, 0) + 1
print(counts)     # {'tea': 2, 'milk': 1}
```

## Background: sorting by part of each item

`sorted(items)` returns a new sorted list. For tuples it sorts by the first part. To sort by something else, pass `key=`: a function that, given one item, returns what to sort by. `lambda pair: pair[1]` is a one-line function meaning "the second part". `reverse=True` puts the biggest first. `[:n]` keeps only the first `n`.

```python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
menu = [("tea", 250), ("cake", 400), ("toast", 150)]
print(sorted(menu))                                       # by name
print(sorted(menu, key=lambda pair: pair[1]))             # by price, cheapest first
print(sorted(menu, key=lambda pair: pair[1], reverse=True))   # by price, dearest first
print(sorted(menu, key=lambda pair: pair[1])[:2])         # the 2 cheapest
scores = {"ann": 7, "ben": 5}
print(list(scores.items()))                               # a dict as a list of (key, value) pairs
```
