## Background: `return` inside a loop

`return` ends the whole function immediately, even in the middle of a loop. So a loop that `return`s when it finds something stops at the **first** match. If the loop finishes without returning, nothing matched: the line after the loop runs, and that's where the default answer goes.

```python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
def first_even(numbers):
    for n in numbers:
        if n % 2 == 0:
            return n          # found one: leave the function now; the rest isn't checked
    return None               # only reached if the loop finished without finding one

print(first_even([3, 7, 8, 10]))   # 8, not 10
print(first_even([3, 7]))          # None
```
