## Background: catching an error

`int("abc")` raises a `ValueError` and would stop the program. `try:` runs some code; if that code raises the error named after `except`, Python jumps to the `except` block instead of crashing. Like `cmd || fallback` in a shell.

```python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
for text in ["3", "abc"]:
    try:
        n = int(text)
        print(text, "->", n)
    except ValueError:
        print(text, "-> not a number, handled instead of crashing")
```
