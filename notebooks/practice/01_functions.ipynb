## How a function is built (worked example)

`def`, a name, the inputs in `()`, and after `->` what it gives back. The docstring says what it
does. `return` hands the answer back to whoever called it; nothing after a `return` runs.

````python
def is_well_known(port: int) -> bool:
    """True if port is below 1024."""
    return port < 1024

print(is_well_known(22), is_well_known(8080))
````

The input name (`port`) is just a label for whatever gets passed in. Calling `is_well_known(22)` runs the body with `port` meaning 22.

## 2. Make a decision

**Receives:** `port`, a number.
**Returns:** `"well-known"` if it's below 1024, `"registered"` if it's below 49152, otherwise `"dynamic"`.

**New tool:** `if` / `elif` / `else`. Each branch can have its own `return`.

**Your task:** replace the `raise` line with an `if` / `elif` / `else` that returns one of the three words.
**Done when:** the check says `port_class: 5/5 passing`.

````python
def port_class(port: int) -> str:
    """"well-known" below 1024, "registered" below 49152, otherwise "dynamic"."""
    if port < 1024:
        return "well-known"
    elif port < 49152:
        return "registered"
    else:
        return "dynamic"
````

````python
check(port_class, [((0,), "well-known"), ((1023,), "well-known"), ((1024,), "registered"),
                   ((49151,), "registered"), ((49152,), "dynamic")]);
````

## 3. Split text into pieces

**Receives:** `hostname`, text such as `"nyc-core-sw-01"`.
**Returns:** the text before the first `-`. So `"nyc-core-sw-01"` gives `"nyc"`, and `"edge"` (no `-`) gives `"edge"`.

**New tool:** `.split("-")` cuts text at every `-` and gives back a **list** of the pieces. To take one
item out of a list by position, write `[position]` after it, counting from 0 (notebook 01 §C: `path[0]`).

**First:** run the **Demo** cell below as it is, and look at what it prints.
**Your task:** in the `def site_code` cell, replace the `raise` line with one `return` line that uses
`hostname.split(...)` and a position.
**Done when:** the check says `site_code: 3/3 passing`.

````python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
pieces = "nyc-core-sw-01".split("-")
print(pieces)              # the list of pieces
print(pieces[1])           # the item at position 1 (positions count from 0)
print("edge".split("-"))   # no "-" in the text: a list with just one item
````

````python
def site_code(hostname: str) -> str:
    """The part before the first "-": "nyc-core-sw-01" -> "nyc". No "-" at all: the whole name."""
    return f"{hostname.split("-")[0]}"
````

## 4. Split, then convert the type

**Receives:** `hostport`, text such as `"10.0.0.5:443"`.
**Returns:** two values, the address as text and the port as a **number**: `("10.0.0.5", 443)`.

**New tools:**
- `.split(":")` cuts at the `:` and gives a list of two pieces.
- `host, port = some_list_of_two` puts the first piece in `host` and the second in `port` (notebook 01 §C unpacked a flow the same way).
- `int("443")` turns text into the number 443.
- `return a, b` returns two values together.

**Your task:** replace the `raise` line with two lines: one that splits and unpacks, one that returns.
**Done when:** the check says `split_hostport: 2/2 passing`.

````python
def split_hostport(hostport: str) -> tuple[str, int]:
    """"10.0.0.5:443" -> ("10.0.0.5", 443). The port comes back as an int."""
    return hostport.split(":")[0], int(hostport.split(":")[1])
    
````

````python
check(split_hostport, [(("10.0.0.5:443",), ("10.0.0.5", 443)), (("8.8.8.8:53",), ("8.8.8.8", 53))]);
````

## 5. Loop and keep a running total

**Receives:** `sizes`, a list of numbers such as `[1532, 74, 60]`.
**Returns:** their total, `1666`. An empty list `[]` gives `0`. Don't use `sum()`; the point is to build it yourself.

**New tool:** a `for` loop. The indented block under it runs once per item, and the loop variable
holds each item in turn:
```python
for size in sizes:
    ...   # runs 3 times for [1532, 74, 60]: size is 1532, then 74, then 60
```
`total = total + size` (or the short form `total += size`) adds to a running total.

**Your task:** replace the `raise` line with: a line that sets a running total to `0`, a `for` loop
that adds each size to it, and a `return` of the total after the loop (not indented under it).
**Done when:** the check says `total_bytes: 3/3 passing`.

````python
def total_bytes(sizes: list[int]) -> int:
    """[1532, 74, 60] -> 1666. An empty list gives 0."""
    total = 0
    for size in sizes:
        total = total + size
    return total
````

````python
check(total_bytes, [(([1532, 74, 60],), 1666), (([],), 0), (([1500],), 1500)]);
````

## 6. Loop over pieces and build a list

**Receives:** `ip`, text such as `"10.0.0.1"`.
**Returns:** a list of four **numbers**: `[10, 0, 0, 1]`.

**New tools:**
- `result = []` makes an empty list.
- `result.append(x)` adds `x` to the end of it.
- `ip.split(".")` gives the four pieces as text, and `int(...)` converts one piece.

**Your task:** replace the `raise` line with: a line that makes an empty list, a `for` loop over
`ip.split(".")` that appends each piece converted to a number, and a `return` of the list after the loop.
**Done when:** the check says `octets: 2/2 passing`.

````python
def octets(ip: str) -> list[int]:
    """"10.0.0.1" -> [10, 0, 0, 1]. Each piece is an int, not text."""
    result = []
    ipsplit = ip.split(".")[0], ip.split(".")[1], ip.split(".")[2], ip.split(".")[3]
    for ips in ipsplit:
        result.append(int(ips))
    return result
````

````python
check(octets, [(("10.0.0.1",), [10, 0, 0, 1]), (("192.168.1.10",), [192, 168, 1, 10])]);
````
