## Background: whole parts and leftovers

`a // b` divides and drops the remainder. `a % b` gives only the remainder. With 125 minutes: `125 // 60` is `2` (2 whole hours) and `125 % 60` is `5` (5 minutes left over). Putting it back together: `2 * 60 + 5` is `125`.

```python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
print(125 // 60, 125 % 60)     # 2 5
print(2 * 60 + 5)              # 125
print(305 // 100, 305 % 100)   # 3 5   (3 dollars, 5 cents)
```

## Background: f-strings

Put `f` before the opening quote, and anything inside `{}` is replaced by its value. A format code after a `:` inside the braces controls how it looks: `:02d` means "a whole number, at least 2 digits, padded with zeros on the left".

```python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
n = 5
print(f"{n} apples")       # 5 apples
print(f"{n:02d}")          # 05
print(f"{12:02d}")         # 12 (already 2 digits, so no padding)
print(f"{7:03d}")          # 007
```

## Background: split and unpack

`text.split("-")` cuts the text at every `-` and returns a **list of text pieces**. If you know how many pieces there are, you can **unpack** them into names in one line: `year, month, day = "2026-03-01".split("-")`. The pieces are still text, so use `int(...)` to get numbers.

```python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
pieces = "2026-03-01".split("-")
print(pieces)                       # ['2026', '03', '01']
print(pieces[0], pieces[1])         # first and second piece
year, month, day = "2026-03-01".split("-")
print(year, type(year).__name__)    # still text: '2026' is a str
print(int(month) + 1)               # int('03') is 3
```
