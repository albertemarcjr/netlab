## Background: a dict of lists

To group things, keep a dict whose values are lists. `groups.setdefault(key, [])` gives you the list stored under `key`, first creating an empty one if the key isn't there yet. Then `.append(x)` adds to that list.

```python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
groups = {}
groups.setdefault("fruit", []).append("apple")    # creates the "fruit" list, then appends
groups.setdefault("fruit", []).append("pear")     # the list exists now; just appends
groups.setdefault("dairy", []).append("milk")
print(groups)
print("apple"[0])                                 # [0] on text is its first letter
```

## Background: copying, and `while`

`b = a` does **not** copy a list: both names point at the same list, so changing `b` changes `a`. `b = list(a)` makes a real, separate copy.

A `while condition:` loop repeats **as long as** the condition is true. `while items and items[-1] == "":` means "while the list isn't empty and its last item is empty text". The `items and` part stops the loop when the list runs out, before `items[-1]` would fail.

```python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
a = [1, 2, 3]
b = a              # NOT a copy: two names for one list
b.pop()
print(a)           # [1, 2]: a changed too!

a = [1, 2, 3]
c = list(a)        # a real copy
c.pop()
print(a, c)        # [1, 2, 3] [1, 2]

items = ["x", "y", "", ""]
while items and items[-1] == "":
    items.pop()
print(items)       # ['x', 'y']
```
