## Background: following a chain

A dict can point from one thing to the next: `{"a": "b", "b": "c", "c": None}` means a leads to b, b leads to c, and c leads nowhere. To walk the chain, start somewhere and keep replacing your position with `chain[position]` until it's `None`. That's a `while position is not None:` loop.

```python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
chain = {"a": "b", "b": "c", "c": None}
position = "a"
while position is not None:
    print("at", position)
    position = chain[position]    # step to whatever this one points to
```

## Background: a changed copy of a dict, in one line

A **comprehension** builds a new list or dict from an old one. `[n for n in names if n != "ben"]` is a new list without ben. `{k: v for k, v in d.items() if k != "ben"}` is a new dict without the key ben. You can put a list comprehension in the value part to change every list at once.

```python
# Demo: run this cell as it is (Shift+Enter). Not an exercise; don't change it.
names = ["ann", "ben", "cy"]
print([n for n in names if n != "ben"])
d = {"ann": 1, "ben": 2}
print({k: v for k, v in d.items() if k != "ben"})
print({k: [x * 10 for x in v] for k, v in {"a": [1, 2]}.items()})
```
